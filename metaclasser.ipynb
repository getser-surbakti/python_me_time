**Metaclass** adalah class pembuat class. Kalau class biasa dipakai untuk membuat objek, maka metaclass dipakai untuk membuat class itu sendiri. Secara default, semua class di Python dibuat oleh metaclass bawaan bernama "type".

- Objek dibuat dari class
- Class sendiri adalah objek, dan dibuat oleh metaclass.
- Metaclas -> Class -> Object

```python
class MyClass:
    pass
print(type(MyClass))
```

=> "MyClass" adalah instance dari "type" (metaclass default)

**Cara Membuat Metaclass** : Metaclass biasanya dibuat dengan mewarisi dari dari "type" dan override "__new__" atau "__init__" .

```python
class Meta(type):
    def __new__(cls, name, bases, dct):
        dct['greet']= lambda self: f"Hello from {name}"
        return super().__new__(cls, name, bases, dct)

class Person(metaclass=Meta):
    def __init__(self,name):
        self.name=name

p = Person("Alice")
print(p.greet())
```
