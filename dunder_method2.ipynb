```python
class Dog:
    def __init__(self,name):
        self.name = name

    def __str__(self):
        return f"Nama dari anjing ini adalah {self.name}"

dog =Dog("Putih")
print(dog)
```

```python
class QuestionBank:
    def __init__(self, questions):
        self.questions = questions

    def __len__(self):
        return len(self.questions)   # dipanggil saat len(obj)

    def __getitem__(self, index):
        return self.questions[index] # dipanggil saat obj[i]

    def __repr__(self):
        return f"<QuestionBank dengan {len(self)} soal>"

bank = QuestionBank(["Soal 1", "Soal 2", "Soal 3"])

print(len(bank))     # 3  → otomatis pakai __len__
print(bank[1])       # Soal 2 → otomatis pakai __getitem__
print(bank)          # <QuestionBank dengan 3 soal> → otomatis pakai __repr__
```

```python
class Demo:
    def __new__(cls, *args, **kwargs):
        print("__new__ dipanggil")
        return super().__new__(cls)

    def __init__(self, value):
        print("__init__ dipanggil")
        self.value = value

    def __del__(self):
        print("__del__ dipanggil")

    # Representasi
    def __str__(self):
        return f"Demo dengan value={self.value}"

    def __repr__(self):
        return f"<Demo value={self.value}>"

    # Operator
    def __add__(self, other):
        return Demo(self.value + other.value)

    def __eq__(self, other):
        return self.value == other.value

    def __lt__(self, other):
        return self.value < other.value

    # Container
    def __len__(self):
        return len(str(self.value))

    def __getitem__(self, index):
        return str(self.value)[index]

    def __iter__(self):
        return iter(str(self.value))

    # Context Manager
    def __enter__(self):
        print("Masuk context")
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        print("Keluar context")
        return False

a = Demo(10)
b = Demo(20)

print(a)          # __str__
print(repr(a))    # __repr__

print(a + b)      # __add__
print(a == b)     # __eq__
print(a < b)      # __lt__

print(len(a))     # __len__
print(a[0])       # __getitem__
for ch in a:      # __iter__
    print(ch)

with Demo(99) as d:   # __enter__ & __exit__
    print(d)
```

```python
class QuestionBank:
    def __init__(self):
        self.questions = []

    def __repr__(self):
        # tampilkan jumlah soal saat objek dicetak
        return f"QuestionBank({len(self.questions)} soal)"

    def __add__(self, question):
        # tambah soal dengan operator +
        self.questions.append(question)
        return self   # supaya bisa chaining

    def __sub__(self, question):
        # hapus soal dengan operator -
        if question in self.questions:
            self.questions.remove(question)
        return self

bank = QuestionBank()

bank + "Soal Matematika"
bank + "Soal Bahasa Indonesia"
print(bank)   # QuestionBank(2 soal)

bank - "Soal Matematika"
print(bank)   # QuestionBank(1 soal)
```
