```python
import matplotlib.pyplot as plt
import numpy as np
import math

my_x =np.linspace(-1*math.pi,2*math.pi)
my_y =np.sin(my_x)
 
plt.plot(my_x,my_y)
title="Plot"
filename="plot.jpg"
plt.title(title)
plt.savefig(filename)
```
