**Metaproramming**: menulis kode yang bisa menghasilkan atau memodifikasi kode lain secara otomatis. Bukan sekadar menjalankan instruksi, tapi kode kamu bisa mencetak, membuat, atau mengubah kode baru.

**Meta** : tentang dirinya sendiri.

```python
#Normal
class Dog:
    def speak(self):
        return"Woof!"

d=Dog()
print(d.speak())
```

```python
#Metaprogramming: bikin class lewat kode
Dog = type("Dog",(),{"speak":lambda self:"Woof!"}) 

d=Dog()
print(d.speak())  
```

```python
class AutoMeta(type):
    def __new__(cls, name, bases, dct):
        dct['info'] = lambda self: f"Class {name} dibuat otomatis"
        return super().__new__(cls,name,bases,dct)

class Test(metaclass=AutoMeta):
    pass

t= Test()
print(t.info())
```

**Decorator**: Kode yang menulis kode lain. Fungsi yang membungkus fungsi lain untuk menambahkan fitur baru tanpa mengubah kode asli. Cara elegan untuk menambahkan fitur ke fungsi/class tanpa harus mengubah isi fungsi aslinya.

```python
def my_decorator(func):
    def wrapper():
        print("Sebelum fungsi dipanggil")
        func()
        print("Setelah fungsi dipanggil")
    return wrapper
@my_decorator
def say_hello():
    print("Hello!")

say_hello()
```

```python
def log(func):
    def wrapper(*args, **kwargs):
        print(f"Memanggil {func.__name__} dengan args={args}")
        return func(*args, **kwargs)
    return wrapper

@log
def tambah(a,b):
    return a+b

print(tambah(2,3))
```

```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start =time.time()
        result =func(*args, **kwargs)
        end = time.time()
        print(f"Fungsi {func.__name__} selesai dalam {end -start:.4f} detik")
        return result
    return wrapper

class ExamSystem:
    @timer
    def validate_questions(self,questons):
        time.sleep(1)
        return all("q" in q for q in questons)

    @timer
    def grade_exam(self,answers):
        time.sleep(2)
        return sum(answers)/len(answers)

exam = ExamSystem()
print(exam.validate_questions([{"q1":"2+5"},{"q2":"3+4"}]))
print(exam.grade_exam([80,90,70,85]))   
```

- Decorator dipakai di Flask/Djanggo untuk routing(@app.route('/home)).
- dipakai di **pytest** untuk testing (@pytest.mark).
- dipakai di **property management** (@property, @classmeethod, @staticmethod).
- membuat kode lebih rapi, reusable, dan modular.
