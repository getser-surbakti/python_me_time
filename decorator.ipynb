```python
def func(function):
    def wrapper():
        print("Starter")
        function()
        print("Ender")

    return wrapper

def say_hello():
    print("Hello!")     
def say_goodbye():
    print("Goodbye!")   

x = func(say_hello)
x()
y = func(say_goodbye)
y()
```

```python
def func(function):
    def wrapper():
        print("Starter")
        function()
        print("Ender")

    return wrapper

def say_hello():
    print("Hello!")     
def say_goodbye():
    print("Goodbye!")   

say_hello  = func(say_hello)
say_hello()
say_goodbye = func(say_goodbye)
say_goodbye()
```

```python
def func(function):
    def wrapper(x):
        print("Starter")
        function(x)
        print("Ender")

    return wrapper

@func
def say_hello():
    print("Hello!") 
@func    
def say_goodbye(x):
    print(x)   

say_hello()
say_goodbye("Getser")
```
