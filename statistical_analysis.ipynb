**1.1**: Knowledge discovery: finding structure in data
==> Tentang bagaimana kita menggali informasi bermakna dari tupukan data yang besar di era informasi modern.

- Perkembangan teknologi telah menghasilkan data yang semakin kompleks.
- Skala data telah berkemmbang pesat dari kilobyte sampa petabyte, exabyte, zettabyte dan yottabyte.
- Big data menawarkan kekayaan innformasi, namun sekaligus potensi disinformasi, memicu kelumpuhan kuantitatif bagi pengambil keputusan.

- Data -> Informasi ->Knowledge ->Wisdom
- Sifat interdisiplinarity:
    a. Ilmu Komputer: Membangun algoritma untuk memanipulasi dan mengorganisasi data
    b. Statistikus & Matematikawan : memberi instruksi mengenai pengembangan dan penerapan metodologi kuantitatif.
    c. Pakar Basis Data : Mengumpulkan dan menyimpan data.
    d. Designer Perangkat Lunak & Insinyur: Menulis program dan membangun perangkat keras untuk menerapkan algoritma analitik.
    e. Pakar Domain/Subjek : menginterpretasi temuan di bidang spesifik.

- Peran Analitik Statistik & Data Mining:
    a. Dasar Analitik: Ilmu statistik dan metode analisis data modern untuk pengumpulan, pendeskripsian, analisis dan interpretassi data.
    b. Pencarian Pola (Data Mining):Usaha untuk menelusuri basis data guna mencari pola respons atau fitur yang menarik dan berharga.
    c. Peran Penting Ahli Domain: Interpretasi yang dibantu pakar domain sangat krusial. Karena terkadang basis data yang besar mengandung fitur struktural yang jika diteliti lebih dalam ternyata hanyalah derau acak (random noise), bukan pola sebaenarnya. Memisahkan pola nyata dari struktur semu seringkali menjadi perpaduan antara seni dan sains.

**1.2** Data quality versus data quantity
=>Tentang pentingnya menjaga mutu data dibandingkan sekadar mengumpulkan data dalam jumlah besar.

- Prinsip Garbage In, Garbage Out
- Distorsi Kualitas Data :
    a. Distorsi tingkat Individu: Kesalahan penempatan titik desimal, digit tertukar, kesalahan pembulatan pengukuran, missing data, kombinasi bidang klasifikasi yang salah.
    b. Distorsi Tingkat Kolektif: Ketika keseluruhan kumpulan data mengalami ketidakteraturan dalam mekanisme pemilihan atau pengambilan sampel: Menggunakan data pasien asma di wilayah yang hangat dan kering untuk menarik kesimpulan mengenai pasien asma di wilayah yang lembab dan dingin, atau adanya bias seleksi (selection bias0 dimana pengambilan sampel secara tidak sadar lebih memilih subjek tertentu saja). Hal ini dapat mencemari basis data dan mengacaukan kerangka sampel sehingga kesimpulan yang ditarik tidak akurat untuk populasi target yang sebenarnya.

- Meskipun memiliki data dalam jumlah besar sangat krusial di era informasi saat ini, menjaga kualitas data melalui validasi, pencegahan bias, dan pengendalian mutu (QA/QC) jauh lebih diutamakan agar proses penemuan pengetahuan (knowledge discovery) dapat menghasilkan wawasan yang benar-benar akurat dan bermakna.

**1.3**Statistical modeling versus statistical description
=> Tentang perbedaan peran antara pemodelan/inferensi statistik dan deskripsi statistik dalam anlaisis data.
- Deskripsi Statistik: Meringkas fitur kuantitatif dan kualitatif dalam suatu sampel atau populasi. Membawa fitur, karakteristik, atau pola tersembunyi dari suatu kelompok data ke permukaan.

- Pemodelan dan Inferensi Statistik: Penarikan kesimpulan mengenai suatu populasi yang lebih luas berdasarkan informasi yang terkandung dalam sampel acak menggunakan teori probabilitas, yang sering kali membutuhkan penerapan model statistik formal.

- Hubungan dan Perbedaan Praktis
    a. Dalam praktiknya, deskripsi statistik dan pemodelan/inferensi sering diterapkan bersamaan. Proses inferensi biasanya memuat elemen-elemen deskriptif di dalamnya.
    b. Dapat juga digunakan terpisahtergantung pada tujuan analisis data. Pendekatan deskriptif untuk investigasi eksploratif, tujuannya hanya mendiskripsikan (disebut EDA = Exploratory Data Analysis)
    c. Sedangkan untuk Pendekatan Inferensial untuk studi perbandingan untuk menentukan apakah terdapat perbedaan pola yang signifikan secara statistik.

- Dalam analitik data modern, analisis eksploratif dan pemodelan statistik saling melengkapi untuk membantu proses penemuan pengetahuan.

**2.1**Concepts in probability
=>Tentang landasan teori probabilitas yang menjadi fondasi utama dalam semua operasi statistik, baik untuk keperluan deskriptif maupun inferensial.

- Data dihasilkan ketika suatu proses acak menghasilkan outcome yang dapat dikuantifikasi atau  dikategorikan.
- Sample Space = Seluruh kemungkinan yang dapat terjadi dari suatu proses acak.
Event = Bagian dari sample space, dapat berupa hasil tunggal maupun gabungan dari beberapa hasil.
- Interpretasi Frekuansi Relatif = Probabilitas suatu event dihitung berdasarkan frekuensi jangka panjang kemunculannya.Jika suatu event diamati secara berulang kali, rasio antara jumlah kemunculan pristiwa tersebut terhadap total percobaan merupakan nilai probabilitasnya.
- Aksioma Probailitas 
    a. Nilai selalu antara 0 dan 1
    b. Total probabilitas dari seluruh ruang sampel adalah 1.
- Aturan Kombinasi Event
    a. Aturan Penjumlahan
    b. AturanKondisionalitas
    c. Aturan Perkalian

- Pemahaman mengenai ruang sampel, peristiwa, serta aturan-aturan probabilitas menjadi dasar logis dan matematis untuk membangun model statistik, melakukan inferensi, serta menggali pengetahuan (knowledge discovery) dari data.

**2.2** Multiple random variables∗
=> Memperluas konsep peuabah acak tunggal (univariate) ke dalam ranah multivariat. Dalam analisis data modern, data hampir selalu melibatkan banyak karakteristik atau variabel secara bersamaan.
- Jika dalam analisis dasar kita biasanya hanya mengamati satu karakteristik saja dari suatu objek (misalnya tinggi badan saja atau berat badan saja), peubah acak berganda (multiple random variables) melibatkan pengamatan dua atau lebih karakteristik sekaligus dari proses acak yang sama (misalnya mencatat tinggi badan, berat badan, tekanan darah, dan usia pasien secara bersamaan).
- Ketika kita berurusan dengan lebih dari satu peubah acak (misalnya dua peubah acak $X$ dan $Y$), perilaku probabilitasnya tidak dapat dianalisis secara terpisah sepenuhnya jika keduanya saling memengaruhi.
    a.Fungsi Distribusi Bersama (Joint Distribution Function): Menyatakan probabilitas gabungan di mana peubah acak X mengambil nilai tertentu dan secara bersamaan peubah acak Y mengambil nilai tertentu lainnya.
    b. Untuk data diskrit dinyatakan sebagai fungsi probabilitas bersama. Sedangkan untuk data kontinu dinyatakan melaluli fungsi kepekatan bersama (joint probability density function).
- Jika kita memiliki distribusi gabungan antara beberapa variabel, kita sering kali ingin mengetahui perilaku probabilitas dari satu variabel saja dengan mengabaikan variabel lainnya.
- Proses matematis untuk "menyingkirkan" variabel lain (biasanya melalui penjumlahan/kondisional untuk data diskrit atau integrasi untuk data kontinu) disebut sebagai pencarian distribusi marginal.
- Independensi Statistik: Dua atau lebih peubah acak dikatakan independen (saling bebas) jika kemunculan nilai dari suatu peubah acak tidak memberikan informasi atau pengaruh apa pun terhadap distribusi nilai peubah acak lainnya.
- Konsep ini sangat krusial dalam analitik data, karena asumsi independensi sering kali mempermudah perhitungan pemodelan statistik, meskipun dalam data dunia nyata (real-world data), variabel-variabel sering kali memiliki tingkat ketergantungan atau korelasi tertentu.

- Dalam praktik analisis data dan data mining, pemahaman tentang peuabah acak berganda menjadi dasar untuk membangun teknik tingkat lanjut seperti:
    a. Analisis multivariate: Regresi berganda, analisis komponen utama(Principal Component Analysis/PCA), dan analisis kluster (cluster analysis).
    b. Memahami struktur hubungan tersembunyi antar-fitur dalam basis data dan berdimensi tinggi.

1. Distribusi Binomial:
    => Tips praktis bagi data analis:
        a. "Apakah data ini bisa dikelompokkan menjadi ya/tidak (sukses/gagal)?" (Jika ya -> Lanjut)
        b. "Apakah ukuran sampelnya sudah pasti sejak awal?" (Jika ya -> Lanjut)
        c. "Apakah kejadian satu sampel mengubah peluang kejadian sampel lainnya?" (Jika tidak -> Memenuhi syarat independen).

Jika ketiga jawaban di atas terpenuhi, maka Anda bisa dengan percaya diri menggunakan Distribusi Binomial untuk memodelkan data tersebut.

2. Distribusi Poisson:
=> Distribusi probabilitas diskrit yang memodelkan jumlah kejadian(events) yang terjadi dalam suatu interval waktu atau ruang tertentu.
    a. Data berupa hitungan diskrit: menyatakan jumlah kejadian (frekuensi kemunculan)dalam suatu unit waktu, luas, atau volume tertentu.
    b. Kejadian bersifat langka
    c. Independensi: kemunculan satu kejadian tidak memengaruhui atau bergantung pada kemunculan kejadian lainnya.
    d. Laju rata-rata konstan.
=> Jika ragu, bisa dilakukan pengujian formal:
    a. Uji rasio varians (Variance-to-Mean Ratio): sifat paling khas-> rata-rata = varians.
    b. Jika nilai varians jauh lebih besar daripada rata-rata (Overdispersion), beralih ke model lain seperti Negative Binomiol Distribution.

3. Distribusi Geometric
=>Ketika mengamati serangkaian percobaan independen (Bernoulli trials) yang masing-masing memiliki hasil biner (sukses atau gagal) dengan ketentuan:
    a. Peluang sukses konstan
    b. Fokus pada jumlah percobaan hingga sukses pertma.
    c. Peubah acak didefinisikan sebagai jumlah percobaan yang dilakukan hingga sebelum kesuksesan pertama.

4. Distribusi Exponential
=> Untuk memodelkan waktu tunggu(waiting times) antar-kejadian atau data masa hidup (lifetime data) dari suatu sistem yang berlanjut secara kontinu.

Karakteristik utama:
    a. Nilai variabel acak selalu positif seperti durasi waktu, jarak atau satuan ukuran kontinu lainnya.
    b. Parameter laju beta menunjukkan seberapa cepat atau sering sebuah pristiwa terjadi.
    c. Sifat tanpa ingatan (memoryless property)

5. Distribusi Gamma
=> Bentuk perluasan dari distribusi eksponensial untuk yang tidak berdistribusi tunggal.
    a. Alpha sebagai parameter bentuk
    b. Betha sebagai parameter skala
    c. Ketika Alpha menyusut menjadi =1, distribusi Gamma menjadi distribusi eksponensial.


6. Distribusi Chi-Square
=>Merupakan kasus khusus dari distribusi Gamma, dimana parameter-parameternya disesuaikan untuk kebutuhan inferensi statistik.
    a. Dikarakterisasi oleh satu parameter utama yaitu derajat kebebasan v (degrees of freedom)
    b. Alpha =v/2 dan Betha=2
    c. Menjadi fondasi penting dalam pengujian hipotesis statistik: Uji kecocokan (goodness-of-fit tests), analisis tabel kontingensi, serta distribusi sampel dari varians sampel pada data berdistribusi normal.


7. Distribusi Normal
=> Salah satu model probabilitas paling penting dan sering digunakan dalam statistika serta data analytics.
    a. Untuk memodelkan Fenomena Kontinu Alami/Sosial: Banyak variabel dunia nyata yang mengelompok di sekitar satu nilai rata-rata (sentral) dengan penurunan frekuensi secara simetris ke arah nilai ekstrem kiri maupun kanan.
    b. Penerapan Central Limit Theoremn, dimana rata-rata dari sampel acak yang cukup besar akan mendekati distribusi normal, terlepas dari apa bentuk distribusi awal populasi asalnya.hal inilah yang menjadikan distribusi normal sebagai fondasi utama dalam uji hipotesis dan inferensi statistik.
    c. Setiap peubah acak dapat distandarisasi ke dalam bentuk distribusi normal standar Z. Proses ini memudahkan analis data untuk menghitung probabilitas menggunakan tabel statistik baku atau fungsi pemerograman komputer seperti "pnorm(z)" dalam lingkungan R.

**Distributions derived from normal**
=> Distribusi normal tidak hanya berdiri sendiri, tetapi juga menjadi fondasi utama (building block) bagi pembentukan berbagai distribusi probabilitas penting lainnya yang sering digunakan dalam inferensi statistik dan data analytics.

**3.1** Random Sampling
=>Dalam analisis data, kita jarang bisa mengukur seluruh anggota dari suatu populasi besar. Oleh karena itu, analis mengambil sebagian kecil anggota populasi yang disebut sampel.

- Konsep i.i.d: Ketika analis data mengambil sampel acak sederhana dari suatu populasi, variabel-variabel yang diamati dari sampel tersebut umumnya diasumsikan bersifat i.i.d. (independent and identically distributed)
- Random sampling penting untuk menghindari Selection Bias, menjadi dasar inferensi Statistik. 

**3.2** Data Types
=> Dalam analisis data, jenis data yang tersedia menentukan model probailitas dan metode statistik apa yang paling tepat digunakan.
    a. Data kuantitatif : mewakili pengukuran numerik, dimana setiap tambahan satu unit pengukuran memiliki makna kuantitatif yang jelas.
    b. Data kualitatif : berdifat kategorikal dan bukan merupakan hasil pengukuran numerik murni.
        Data kualitatif sering kali dikuantifikasi untuk keperluan komputasi komputer, misalnya dengan memberikan label angka seperti accepted =1 atau rejected=0. Namun analis data harus berhati-hati: sekadar melabeli data kualitatif dengan angka tidak serta-merta menjadikannya data kuantitatif yang sesungguhnya, karena operasi aritmatika pada label tersebut tidak selalu bermakna secara matematis.
    
    c. Skala Pengukuran Data Kualitatif & Kuantitaif:
        1. Skala Nominal: tanpa urutan pringkat.
        2. Skala Ordinal: memiliki urutan atau pringkat yang alami, meskipun selisih jarak antar tingkatan tidak dapat diukur secara eksak.
        3. Skala Interval & Rasio (Data Kuantitatif): Pengukuran numerik yang memiliki jarak konsisten(interval) serta titik nol mutlak yang bermakna (rasio), memungkinkan analis untuk menerapkan operasi matematika dan analis statistik tingkat lanjut secar penuh.
